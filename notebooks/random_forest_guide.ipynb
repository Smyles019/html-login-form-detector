# Random Forest Guide for HTML Login Form Detection
Welcome! This guide is an educational walkthrough designed for our team. It explains the end-to-end pipeline of our Random Forest model, from understanding HTML signatures to model evaluation.

## 1. Project Overview & HTML Signatures
### What is an HTML signature?
An HTML signature is a compressed structural representation of a webpage. Instead of looking at the actual text or images on a page, we look at the HTML tags (like `(form`, `(input`, `(div`). This allows us to detect malicious login forms based purely on their structural footprint.

### Why are repeated signatures kept?
During our data exploration, we found repeated HTML signatures. It is crucial to understand that these are **not exact duplicate samples** (since they have unique SHA-256 hashes). Different web pages can have the exact same HTML structure. Removing them would destroy legitimate observations and harm the model's ability to recognize common login structures.

## 2. Preventing Data Leakage
Data leakage happens when information from outside the training dataset is used to create the model. In our pipeline, we ensure that operations like scaling (`RobustScaler`) and tokenization (`CountVectorizer`) are **only fitted on the training data**, and then applied (`transform`) to the test data. This strictly simulates a real-world scenario where the model sees entirely unseen data.

## 3. HTML Structural Features vs N-Grams
Our feature extraction combines two methods:
1. **Structural Features**: We manually extract counts (e.g., number of inputs, forms) and relational metrics (e.g., maximum nesting depth, input-to-div ratio).
2. **HTML N-Grams**: We treat HTML tags as sequential tokens (like `(form` and `)`). Using a `CountVectorizer` with 1- to 3-grams allows the model to learn structural sequences such as `(form(div(input))`.

## 4. What is a Random Forest?
A Random Forest is a machine learning model built on an ensemble (a group) of many **Decision Trees**. 
Instead of relying on a single tree to classify a page as `LOGIN_FORM` or `NO_FORM`, the Random Forest trains hundreds of slightly different trees. Each tree makes a prediction, and the final prediction is decided by majority voting.

**Key Parameters we use:**
- `n_estimators=300`: We build 300 different decision trees.
- `random_state=42`: Ensures our experiment is fully reproducible.
- `class_weight="balanced"`: Adjusts the model to pay more attention to minority classes, handling class imbalance naturally.

## 5. Evaluation Metrics
When we evaluate the model, we look at several metrics:
- **Accuracy**: The percentage of totally correct predictions.
- **Precision**: When the model predicts `LOGIN_FORM`, how often is it actually a login form?
- **Recall**: Out of all the actual login forms, how many did the model successfully find? This is critical for security, because a **False Negative** (missing a login form) can be dangerous.
- **F1-score**: The harmonic mean of Precision and Recall.
- **ROC-AUC**: Measures the model's ability to distinguish between the two classes.
