### 1. The Structure of Our Code

To make things easy to understand and test, the code has been broken down into three logical steps located in the `src/` folder:

1. **`data_processing.py`**: This script does the "dirty work." It loads the raw CSV files, handles missing values, and turns raw HTML signatures into numbers that a computer can understand (like counting how many `<form>` tags there are).
2. **`train_model.py`**: This script takes the clean data, splits it into two piles (one pile for the model to *learn* from, and a smaller pile kept hidden to *test* the model later). It trains the Logistic Regression baseline model and saves it.
3. **`evaluate_model.py`**: This is the script you will use the most! It loads the hidden test data and the trained model, and tells you exactly how well the model performed using plain-English explanations.

### 3. How to Explain the Results to Your Group

When you run `evaluate_model.py`, it will print out several metrics. Here is how you can interpret and explain them:

#### A. The Confusion Matrix
The confusion matrix is just a 2x2 grid that shows where the model got things right and where it got confused.
- **True Positives (TP):** The model caught a phishing page successfully. (YAY!)
- **True Negatives (TN):** The model looked at a normal page and left it alone. (YAY!)
- **False Positives (FP):** The model panicked and flagged a normal page as phishing. (Oops... this means legitimate users might get blocked from logging into their accounts).
- **False Negatives (FN):** The model missed a phishing page. (DANGER! This means an attacker successfully sneaked past our system).

*How to explain it:* "Our model caught [TP] phishing forms, but it completely missed [FN] of them. However, it also accidentally flagged [FP] normal pages, which we need to minimize so we don't annoy real users."

#### B. Precision and Recall
- **Precision:** "Out of all the pages our model *claimed* were phishing, how many actually were?" If precision is low, our system cries "wolf" too often.
- **Recall:** "Out of all the *actual* phishing pages that existed in the test, what percentage did we catch?" If recall is low, our system is blind to many attacks.
- **F1-Score:** This is just a balanced average of Precision and Recall. It's a great single number to look at to see if the model is good overall.

#### C. Feature Importance (What is the model thinking?)
Machine Learning isn't magic. The Logistic Regression model assigns a "weight" (or importance) to every single HTML tag or pattern it sees.

If a feature has a **HIGH POSITIVE weight**, the model thinks: *"When I see this tag, it's highly likely to be a phishing attack!"*
If a feature has a **HIGH NEGATIVE weight**, the model thinks: *"When I see this tag, it's probably a safe, benign page."*

*How to explain it:* "We looked under the hood of the model, and it turns out that phishing pages often have a very specific structure. For example, our model learned that if a page has a high number of [Feature X], it's very likely to be malicious. This proves that attackers are indeed reusing HTML templates!"
